Você trabalha em uma empresa de telecon e tem clientes de vários serviços diferentes, entre os principais, internet e telefone.

O problema é que analisando o histórico dos clientes dos últimos anos, percebeu-se que a empresa está com Churn de mais de 26% dos clientes.

Isso representa uma perda de milhões para a empresa.

O que a empresa precisa fazer para resolver isso ?

```python
import pandas as pd
```

```python
# PASSO 01: Importar a base de dados

tabela = pd.read_csv(r'C:\Python-JupyterNotebook\Hashtag\AutomacaoProcessos\Aula02\Planilha.csv')
```

```python
# PASSO 02: Visualizar a base de dados
 # Entender as informações que há disponível
 # Descobrir os erros da base de dados

display(tabela)
```

```python
# Exclusão da coluna: Unnamed: 0

tabela = tabela.drop('Unnamed: 0', axis = 1) # axis = 0 é linha, axis = 1 é coluna
display(tabela)
```

```python
# PASSO 03: Tratar os dados

display(tabela.info()) # Informações sobre os dados
```

```python
# Resolver os valores que estão sendo reconhecidos de forma errada

tabela['TotalGasto'] = pd.to_numeric(tabela['TotalGasto'], errors = 'coerce') # Força a conversão de texto para número 
display(tabela.info()) # 10 registros estavam como texto
```

```python
# Resolver valores vazios
    # Colunas que tem TODOS os valores vazios, será excluido

tabela = tabela.dropna(how = 'all', axis = 1) # Exclui todos os valores que estão vazios
```

```python
# Resolver valores vazios
    # Linhas que contém pelo menos 1 valor vazio, será excluído

tabela = tabela.dropna(how = 'any', axis = 0) # Exclui a linha que contém valor vazio
```

```python
display(tabela.info())
```

```python
# PASSO 04: Análise Inicial
    # Contar quantos clientes cancelaram e quantos estão ativos (dados da coluna churm)

display(tabela['Churn'].value_counts())
```

```python
# PASSO 04: Análise Inicial
    # Contar quantos clientes cancelaram e quantos estão ativos em porcentagem (dados da coluna churm)

display(tabela['Churn'].value_counts(normalize = True))
```

```python
# PASSO 04: Análise Inicial
    # Contar quantos clientes cancelaram e quantos estão ativos em porcentagem (dados da coluna churm)

display(tabela['Churn'].value_counts(normalize = True).map('{:.2%}'.format))
```

```python
# PASSO 05: Análise detalhada - descobrir as causas do cancelamento
    # Comparar cada coluna da base de dados com a coluna CHURN

import plotly.express as px
```

```python
# Cria o gráfico

coluna = 'Aposentado'
grafico_aposentado = px.histogram(tabela, x = coluna, color = 'Churn')
```

```python
# Cria o gráfico

coluna = 'TipoContrato'
grafico_tipo_contrato = px.histogram(tabela, x = coluna, color = 'Churn', text_auto = True)
```

```python
# Cria o gráfico

coluna = 'MesesComoCliente'
grafico_meses_cliente = px.histogram(tabela, x = coluna, color = 'Churn', text_auto = True)
```

```python
# Cria o gráfico

coluna = 'Genero'
grafico_genero = px.histogram(tabela, x = coluna, color = 'Churn', text_auto = True)
```

```python
# Exibe o gráfico

grafico_aposentado.show()
```

```python
# Exibe o gráfico

grafico_tipo_contrato.show()
```

```python
# Exibe o gráfico

grafico_meses_cliente.show()
```

```python
# Exibe o gráfico

grafico_genero.show()
```

```python
# Criar vários gráficos com todas as colunas da tabela

for coluna in tabela.columns:
    grafico = px.histogram(tabela, x = coluna, color = 'Churn', text_auto = True)
    grafico.show()
```
