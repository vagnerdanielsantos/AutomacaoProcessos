### Trabalhamos em uma importadora e o preço dos nossos produtos é vinculado a cotação de:
    * Dólar
    * Euro
    * Ouro

### Precisamos pegar na internet de forma automática a cotação desses 3 tipos e saber quanto devemos cobrar pelos nossos produtos, considerando uma margem de contribuição que temos na nossa base de dados.

### Para isso, será criado uma automação web:
    * Será utilizado o "selenium"
    * Importante: Será baixado o "webdriver"

```python
# Importar a biblioteca selenium

## Google Chrome --> chromedriver
## Mozilla Firefox --> geockodriver

!pip install selenium
```

```python
# Importar as bibliotecas webdriver

from selenium import webdriver
from selenium.webdriver.common.keys import Keys
```

```python
# Cria o navegador controlado pelo selenium

navegador = webdriver.Chrome()
```

```python
# PASSO 01: Pegar a cotação do Dólar

# Entrar no google chrome
navegador.get('https://www.google.com/')

# Encontra um elemento da tela (campo de digitação da pesquisa)
navegador.find_element('xpath', '//*[@id="APjFqb"]').send_keys('Cotação Dólar Hoje')

# Apertar o Enter
navegador.find_element('xpath', '//*[@id="APjFqb"]').send_keys(Keys.ENTER)

# Encontar o xpath onde contém o valor do Dólar
cotacao_dolar = navegador.find_element('xpath', '//*[@id="knowledge-currency__updatable-data-column"]/div[1]/div[2]/span[1]').get_attribute('data-value')
```

```python
# PASSO 02: Pegar a cotação do Euro

# Entrar no google chrome
navegador.get('https://www.google.com/')

# Encontrar um elemento da tela (campo de digitação da pesquisa)
navegador.find_element('xpath', '//*[@id="APjFqb"]').send_keys('Cotação Euro Hoje')

# Apertar o Enter
navegador.find_element('xpath', '//*[@id="APjFqb"]').send_keys(Keys.ENTER)

# Encontrar o xpath onde contém o valor do Euro
cotacao_euro = navegador.find_element('xpath', '//*[@id="knowledge-currency__updatable-data-column"]/div[1]/div[2]/span[1]').get_attribute('data-value')
```

```python
# PASSO 03: Pegar a cotação do Ouro

# Entrar no site consultar valor do ouro
navegador.get('https://www.melhorcambio.com/ouro-hoje')

# Encontrar o xpath onde contém o valor do Ouro
cotacao_ouro = navegador.find_element('xpath', '//*[@id="comercial"]').get_attribute('value')

navegador.quit()
```

```python
# Exibir as três cotações

print('A cotação do dólar é: R$ ', cotacao_dolar)
print('A cotação do euro é: R$ ', cotacao_euro)
print('A cotação do ouro é R$ ', cotacao_ouro)
```

```python
# Importar a biblioteca Pandas

import pandas as pd
```

```python
# PASSO 04: Importar a base de dados e atualizar a base

tabela = pd.read_excel('C:\Python-JupyterNotebook\Hashtag\AutomacaoProcessos\Aula03\Produtos.xlsx')
```

```python
# Exibir a base de dados sem a atualização ainda

display(tabela)
```

<font color = 'green'><font size = 6>Agora será atualizado os preços e o valor final dos produtos</font></font>

```python
# PASSO 05: Recalcular os preços

# Atualizar a cotação

## Cotação do Dólar
tabela.loc[tabela['Moeda'] == 'Dólar', 'Cotação'] = float(cotacao_dolar)
## Cotação do Euro
tabela.loc[tabela['Moeda'] == 'Euro', 'Cotação'] = float(cotacao_euro)
## Cotação do Ouro
tabela.loc[tabela['Moeda'] == 'Ouro', 'Cotação' = float(cotacao_ouro)
```

```python
display(tabela) # coluna moeda atualizada
```

```python
# Recalcular os preços

## Preço de compra  = preço original * cotação
tabela['Preço de Compra'] = tabela['Preço Original'] * tabela['Cotação']

## preço da venda = preço de compra * margem
tabela['Preço de Venda'] = tabela['Preço de Compra'] * tabela['Margem']
```

```python
# Exibe a tabela atualizada

display(tabela)
```
